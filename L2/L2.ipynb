This primer covers two features a deep agent provides that a basic tool-calling agent does not: a filesystem, and instructions loaded from files. Each section builds the smallest agent to demonstrate the feature in action.

````python
# Setup and imports
import warnings
warnings.filterwarnings('ignore')

from pathlib import Path

from IPython.display import display, Markdown
from deepagent_helpers import langsmith_status
from visualization_helpers import render_report
from deepagents import create_deep_agent
from deepagents.backends import FilesystemBackend
from dotenv import load_dotenv
from langchain.agents.middleware import ModelCallLimitMiddleware
from langchain_core.messages import HumanMessage


load_dotenv(override=True)

MODEL = "openai:gpt-5.4-mini"
ROOT = "primer_workspace"      # the agent's filesystem root


def clear_poems():
    """Remove poems from an earlier run, so the output below is from this one.    """
    for poem in Path(ROOT, "poems").glob("*.md"):
        poem.unlink()


clear_poems()

print("✓ Environment ready")
````

## The Filesystem

A deep agent gets file tools by default: `read_file`, `write_file`, `edit_file`,
`ls`, `glob`, `grep`, and `delete`. Deep agents supports a variety of backends
including persistent databases and the local filesystem. For this course you
will use the local filesystem.

`FilesystemBackend(root_dir=..., virtual_mode=True)` is what points those tools
at a directory. The agent sees `root_dir` as `/`, and paths above it are
unreachable, so `/skills/haiku/SKILL.md` means
`primer_workspace/skills/haiku/SKILL.md`, and nothing outside that directory is
in range.

The cell below creates the backend and lists the files the agent can see.

````python
backend = FilesystemBackend(root_dir=ROOT, virtual_mode=True)

print("\nwhat the agent sees as its filesystem:")
for f in sorted(Path(ROOT).rglob("*")):
    if f.is_file():
        print("   /" + str(f.relative_to(ROOT)))
````

## Instructions: AGENTS.md and Skills

Instructions can reach a deep agent in three ways. They differ in **where** you specify them and **when** they load:

| | Where it's specified | When it loads | Best for |
|---|---|---|---|
| `system_prompt` | In code, when the agent is created | Every time the agent is called | A short, fixed instruction |
| `memory=["/memories/AGENTS.md"]` | A file the agent can read | At the start of every run | What is always true: role, procedure, rules |
| `skills=["/skills/"]` | One `SKILL.md` per skill, in its own directory | Description up front, full instructions only when needed | Specialized procedures the agent needs only sometimes |

### AGENTS.md: always on

`memory=["/memories/AGENTS.md"]` loads that file into the prompt at the start of every run. Use it for what is always true: the agent's role, the procedure it follows, the rules it must not break. It costs context on every call, so it should stay short. In this lesson, it specifies that the agent is a family-friendly poet.

### Skills: progressive disclosure

`skills=["/skills/"]` works by *progressive disclosure*. Each skill is a directory holding a `SKILL.md` with YAML frontmatter:

```
---
name: haiku
description: Write a haiku, a three line poem with a 5-7-5 syllable pattern...
---

(the full instructions follow)
```

Only the `name` and `description` are loaded up front. The agent reads the body with `read_file` when it decides a skill applies. So twenty skills cost twenty lines of context, not twenty documents, and the agent pulls in the detail for the one it needs.

### Our poet's workspace

Our poet has one `AGENTS.md` and four skills:

```
primer_workspace/
  memories/AGENTS.md          "You are a poet that has skills..."
  skills/haiku/SKILL.md
  skills/limerick/SKILL.md
  skills/sonnet/SKILL.md
  skills/acrostic/SKILL.md
  poems/                      output
```

This layout may look familiar. Claude Code, Cursor, and other coding assistants organize instructions the same way, for the same reason: the always-on rules go in one file, and the specialized procedures sit behind a description the agent reads only when it matters.

##### Build the poet

One call, `create_deep_agent`, wires everything together:
- `model`: the language model doing the work
- `system_prompt`: deliberately tiny. It only points the agent to `/memories/AGENTS.md`
- `backend`: the filesystem from above, so the file tools work inside `primer_workspace`
- `memory`: files loaded into every run (here, `AGENTS.md`)
- `skills`: the directory of skills to publish by description
- `middleware`: `ModelCallLimitMiddleware(run_limit=30, ...)` is a safety cap that ends a run after 30 model calls, so a confused agent can't loop forever. It is a similar idea to `MAX_TURNS` in L3.

````python
poet = create_deep_agent(
    model=MODEL,
    system_prompt="Your instructions are in /memories/AGENTS.md.",
    backend=backend,
    memory=["/memories/AGENTS.md"],
    skills=["/skills/"],
    middleware=[ModelCallLimitMiddleware(run_limit=30, exit_behavior="end")],
)
print("\u2713 Poet agent built")
````

##### Run the Poet

Now we give the poet a task that needs all four skills: write a Red Sox poem in each form (haiku, limerick, sonnet, and an acrostic on BOSTON), each saved to its own file under `/poems/`.

After the agent finishes, the cell does three things:
1. Prints the agent's **tool calls, in order**, taken from its message history
2. Reads the saved poems back from the `/poems/` folder
3. Displays each poem

Poems from any earlier run were deleted during setup, so what you see comes from this run.

As you read the tool calls, look for the `read_file` calls on `SKILL.md` files. The agent started with only each skill's short description, so these calls are **progressive disclosure** happening: it loads a skill's full instructions only when it decides it needs them.

````python
request = HumanMessage(content=(
    "Write a poem about the Boston Red Sox in each of your forms: haiku, "
    "limerick, sonnet, and an acrostic on the word BOSTON. Write each poem to "
    "its own file under /poems/, named after the form."
))

result = poet.invoke(
    {"messages": [request]},
    config={"run_name": "primer / poet, four forms",
            "tags": ["deepagents-primer", "skills"]},
)

# The read_file calls on SKILL.md are progressive disclosure happening: the
# agent pulled in each form's rules only when it came to write that form.
print("tool calls, in order:\n")
for m in result["messages"]:
    for c in (getattr(m, "tool_calls", None) or []):
        print(f"  {c['name']:12} {c['args'].get('file_path', '')}")

for poem in sorted(Path(ROOT, "poems").glob("*.md")):
    # Two trailing spaces force a Markdown line break: a poem has no blank
    # lines between verses, so without this every line would collapse into
    # one run-on paragraph.
    text = poem.read_text().strip().replace("\n", "  \n")
    display(render_report(text, title=f"/poems/{poem.name}"))

print("\n" + langsmith_status())
````

## Conclusion

Two features, each one line of configuration:

- **A filesystem.** `backend=FilesystemBackend(root_dir=..., virtual_mode=True)`
  gives the agent file tools scoped to one directory, and makes its output
  something you can open rather than something you have to parse out of a
  message.
- **Instructions as files.** `memory=` loads what is always true on every run.
  `skills=` publishes a description per skill and lets the agent read the full
  procedure only when it applies, so a large library stays cheap.

**Key takeaways**
- **Choose where an instruction lives by when it is needed.** Always-true rules go in `AGENTS.md`. Specialized procedures go in skills, so they only cost context when used.
- **Progressive disclosure keeps context small.** Twenty skills cost twenty short descriptions, not twenty full documents.
- **Guardrails are one line.** `ModelCallLimitMiddleware` caps how long a run can go.
- **Tool calls show what happened.** Reading them, or the LangSmith trace, tells you which instructions the agent loaded and when.

**Try it yourself**
- Change the request, for example ask for only a haiku, and see which skills the agent reads
- Edit a `SKILL.md`, such as the haiku rules, then rerun and see how the poem changes
- Add a new skill directory with its own `SKILL.md`, rebuild the agent, and ask for that form
